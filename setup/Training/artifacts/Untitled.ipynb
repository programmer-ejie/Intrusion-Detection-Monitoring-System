```python
import joblib, pandas as pd

BUNDLE = joblib.load("artifacts/ids_ensemble.joblib")  # use the NEW file
model = BUNDLE["bin"]
features = BUNDLE["features"]

# load one row from attack.csv that is labeled Bot
df = pd.read_csv("attack.csv", low_memory=False)
df.columns = [c.strip().lower().replace(" ", "_").replace("/s","_s").replace(".","_") for c in df.columns]
row = df[df["label"].str.strip().str.upper() == "BOT"].iloc[[0]]

# keep only expected features
X = row.reindex(columns=features, fill_value=0)

proba = model.predict_proba(X)[:,1][0]
print("prob_attack:", proba)
print("thr:", BUNDLE["best_threshold"])
```
