```python
import joblib
import numpy as np
import pandas as pd
import re
import unicodedata
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Optional, Dict, Any, Tuple


# ---------- Utilities ----------
def normalize_colname(name: str) -> str:
    s = str(name).strip()
    s = s.replace("%", "pct")
    s = re.sub(r"[\/\\\s\-\.\(\)]+", "_", s)
    s = re.sub(r"_+", "_", s).strip("_").lower()
    return s

def clean_label(s: str) -> str:
    s = unicodedata.normalize("NFKC", str(s))
    s = s.replace("–", "-").replace("—", "-")
    s = re.sub(r"\s+", " ", s)
    return s.strip().upper()

def safe_numeric(df: pd.DataFrame) -> pd.DataFrame:
    # convert everything to numeric where possible; non-numeric -> NaN -> 0
    out = df.copy()
    for c in out.columns:
        out[c] = pd.to_numeric(out[c], errors="coerce")
    out = out.replace([np.inf, -np.inf], np.nan).fillna(0.0)
    return out


# ---------- Result structure ----------
@dataclass
class IDSDecision:
    prob_attack: float
    thr: float
    is_attack: bool
    risk: str
    attack_family: Optional[str] = None
    attack_family_proba: Optional[float] = None


# ---------- Core Inference Class ----------
class IDSInference:
    def __init__(self, bundle_path: str = "artifacts/ids_ensemble.joblib"):
        self.bundle = joblib.load(bundle_path)
        self.model_bin = self.bundle["bin"]
        self.model_mul = self.bundle.get("mul", None)
        self.features = list(self.bundle["features"])
        self.thr = float(self.bundle["best_threshold"])

        # Optional "risk bands" for SOC-like behavior
        # (does NOT change your final attack decision threshold)
        self.risk_bands = [
            (0.999, "CRITICAL"),
            (0.95,  "HIGH"),
            (0.80,  "MEDIUM"),
            (0.50,  "LOW"),
            (0.00,  "BENIGN"),
        ]

    def _risk_label(self, p: float) -> str:
        for cut, name in self.risk_bands:
            if p >= cut:
                return name
        return "BENIGN"

    def _prepare_X(self, df_raw: pd.DataFrame) -> pd.DataFrame:
        # Normalize columns exactly like training
        df = df_raw.copy()
        df.columns = [normalize_colname(c) for c in df.columns]

        # Keep only expected features; fill missing with 0
        X = df.reindex(columns=self.features, fill_value=0.0)

        # Make sure numeric and finite
        X = safe_numeric(X)

        return X

    def predict_df(self, df_raw: pd.DataFrame) -> pd.DataFrame:
        """
        Returns a dataframe with:
        prob_attack, thr, is_attack, risk, attack_family, attack_family_proba
        """
        X = self._prepare_X(df_raw)
        prob_attack = self.model_bin.predict_proba(X)[:, 1]
        is_attack = prob_attack >= self.thr
        risk = [self._risk_label(float(p)) for p in prob_attack]

        # Stage-B: only for those flagged as attack by Stage-A
        attack_family = np.array([None] * len(df_raw), dtype=object)
        attack_family_proba = np.array([np.nan] * len(df_raw), dtype=float)

        if self.model_mul is not None and is_attack.any():
            idx = np.where(is_attack)[0]
            X_atk = X.iloc[idx]
            fam = self.model_mul.predict(X_atk)

            # If classifier supports predict_proba, capture top confidence
            fam_p = None
            if hasattr(self.model_mul, "predict_proba"):
                proba_mul = self.model_mul.predict_proba(X_atk)
                fam_p = proba_mul.max(axis=1)

            attack_family[idx] = fam
            if fam_p is not None:
                attack_family_proba[idx] = fam_p

        out = pd.DataFrame({
            "prob_attack": prob_attack.astype(float),
            "thr": self.thr,
            "is_attack": is_attack.astype(bool),
            "risk": risk,
            "attack_family": attack_family,
            "attack_family_proba": attack_family_proba,
        })
        return out

    def predict_one(self, row: pd.Series) -> IDSDecision:
        df = pd.DataFrame([row])
        out = self.predict_df(df).iloc[0].to_dict()
        return IDSDecision(
            prob_attack=float(out["prob_attack"]),
            thr=float(out["thr"]),
            is_attack=bool(out["is_attack"]),
            risk=str(out["risk"]),
            attack_family=out["attack_family"],
            attack_family_proba=(None if pd.isna(out["attack_family_proba"]) else float(out["attack_family_proba"])),
        )


# ---------- BOT batch test + plot ----------
def batch_test_bot(attack_csv_path: str, bundle_path: str = "artifacts/ids_ensemble.joblib") -> Tuple[pd.DataFrame, pd.DataFrame]:
    ids = IDSInference(bundle_path=bundle_path)

    df = pd.read_csv(attack_csv_path, low_memory=False)
    df.columns = [normalize_colname(c) for c in df.columns]

    if "label" not in df.columns:
        raise ValueError("attack.csv has no 'label' column after normalization.")

    label_norm = df["label"].astype(str).map(clean_label)
    bot_mask = label_norm.eq("BOT")

    if not bot_mask.any():
        # show what labels exist to help debugging
        counts = label_norm.value_counts().head(30)
        raise ValueError(f"No BOT samples found. Top labels:\n{counts}")

    bot_df = df.loc[bot_mask].copy()
    results = ids.predict_df(bot_df)

    merged = pd.concat([bot_df.reset_index(drop=True), results.reset_index(drop=True)], axis=1)

    # Summary stats
    summary = pd.DataFrame({
        "count": [len(results)],
        "min_prob": [results["prob_attack"].min()],
        "mean_prob": [results["prob_attack"].mean()],
        "median_prob": [results["prob_attack"].median()],
        "max_prob": [results["prob_attack"].max()],
        "flagged_attack_count": [int(results["is_attack"].sum())],
        "threshold": [ids.thr],
    })

    return merged, summary

def plot_probability_distribution(probs: np.ndarray, title: str = "Probability distribution"):
    plt.figure()
    plt.hist(probs, bins=50)
    plt.title(title)
    plt.xlabel("prob_attack")
    plt.ylabel("count")
    plt.show()
```

```python
from ids_infer import batch_test_bot

bot_merged, bot_summary = batch_test_bot("attack.csv", "artifacts/ids_ensemble.joblib")
print(bot_summary)
print(bot_merged[["prob_attack", "thr", "is_attack", "risk", "attack_family", "attack_family_proba"]].head(10))
```
