```python
import os, glob, re, unicodedata, joblib
import numpy as np
import pandas as pd

# ================= CONFIG =================
DATA_DIR = "Dataset"
BUNDLE_PATH = "artifacts/ids_ensemble.joblib"

# Dual-gate thresholds (do NOT change your trained CRITICAL threshold)
SUSP_THR  = 0.90
WATCH_THR = 0.70

# ================= UTIL =================
def normalize_colname(name):
    s = str(name).strip()
    s = s.replace("%", "pct")
    s = re.sub(r"[\/\\\s\-\.\(\)]+", "_", s)
    s = re.sub(r"_+", "_", s).strip("_").lower()
    return s

def clean_label(s):
    s = unicodedata.normalize("NFKC", str(s))
    s = s.replace("–", "-").replace("—", "-")
    s = re.sub(r"\s+", " ", s)
    return s.strip().upper()

def safe_numeric(df):
    for c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")
    return df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

# ================= LOAD BUNDLE =================
BUNDLE = joblib.load(BUNDLE_PATH)
model_bin = BUNDLE["bin"]
model_mul = BUNDLE.get("mul", None)
features  = BUNDLE["features"]
CRIT_THR  = float(BUNDLE["best_threshold"])

print("Loaded bundle")
print("CRITICAL threshold:", CRIT_THR)
print("SUSPICIOUS threshold:", SUSP_THR)
print("WATCH threshold:", WATCH_THR)

# ================= LOAD ALL CSVs =================
files = sorted(glob.glob(os.path.join(DATA_DIR, "*.csv")))
if not files:
    raise SystemExit(f"No CSVs found in {DATA_DIR}")

dfs = []
for f in files:
    d = pd.read_csv(f, low_memory=False)
    d.columns = [normalize_colname(c) for c in d.columns]
    d["capture_id"] = os.path.basename(f).replace(".csv", "")
    dfs.append(d)

df = pd.concat(dfs, ignore_index=True)

# keep true labels if present (optional)
label_col = next((c for c in df.columns if "label" in c.lower()), None)
if label_col:
    df["true_label"] = df[label_col].astype(str).map(clean_label)

# ================= PREPARE FEATURES =================
X = df.reindex(columns=features, fill_value=0.0)
X = safe_numeric(X)
X = X.clip(lower=-1e12, upper=1e12)  # extra hardening

# ================= STAGE A: PROB + SEVERITY =================
df["prob_attack"] = model_bin.predict_proba(X)[:, 1]

df["severity"] = np.select(
    [
        df["prob_attack"] >= CRIT_THR,
        df["prob_attack"] >= SUSP_THR,
        df["prob_attack"] >= WATCH_THR,
    ],
    ["CRITICAL", "SUSPICIOUS", "WATCH"],
    default="BENIGN"
)

alerts = df[df["severity"] != "BENIGN"].copy()

print("\n===== SUMMARY =====")
print("Total flows:", len(df))
print("Alerts (non-benign):", len(alerts))
print(alerts["severity"].value_counts(dropna=False))

print("\nMax prob_attack:", df["prob_attack"].max())
print("Mean prob_attack:", df["prob_attack"].mean())

# ================= STAGE B: FAMILY CLASSIFICATION (GATED) =================
# In Option B, it's reasonable to run Stage-B for SUSPICIOUS+WATCH too.
# If you want Stage-B only for CRITICAL, replace alerts with alerts[alerts["severity"]=="CRITICAL"].
if model_mul is not None and len(alerts) > 0:
    X_alert = alerts.reindex(columns=features, fill_value=0.0)
    X_alert = safe_numeric(X_alert)
    X_alert = X_alert.clip(lower=-1e12, upper=1e12)

    alerts["predicted_family"] = model_mul.predict(X_alert)

    if hasattr(model_mul, "predict_proba"):
        fam_proba = model_mul.predict_proba(X_alert)
        alerts["family_confidence"] = fam_proba.max(axis=1)
    else:
        alerts["family_confidence"] = np.nan
else:
    alerts["predicted_family"] = None
    alerts["family_confidence"] = np.nan

# ================= SAVE OUTPUT =================
out_file = "ids_alerts_dual_gate.csv"
alerts.to_csv(out_file, index=False)
print(f"\nSaved → {out_file}")

# Show a quick preview
show_cols = ["capture_id", "prob_attack", "severity", "predicted_family", "family_confidence"]
if "true_label" in alerts.columns:
    show_cols.insert(2, "true_label")
print("\nPreview:")
print(alerts[show_cols].head(20))
```
